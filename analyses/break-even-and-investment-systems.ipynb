# Break-Even Analysis and a Three-Variable Investment System

**1)** A shoe maker has fixed costs of \$50,000 and a variable cost of \$18 per pair, and sells each pair for \$130. Find the break-even point.

```python
import matplotlib.pyplot 
from matplotlib.pyplot import *
import numpy 
from numpy import linspace
```

```python
x= linspace(0,500,100)
y= 18*x+50000
z= 130*x
```

```python
figure()  
plot (x,y)
plot (x,z)
legend (('cost','revenue'),loc=2)
title ('Breakeven Analysis')
show()
```

C(x) = 50000 + 18x

R(x) = 130x

50000 + 18x = 130x

50000 = 112x

x = 50000/112 ≈ 446.43

R(446.43) = 130 × 446.43 ≈ 58,035.71

The breakeven point occurs at about (446.43, 58,035.71). 447 pairs of shoes would need to be sold to break even.

```python
breakeven_x = 50000 / (130 - 18)
print(f"Break-even quantity: {breakeven_x:.2f} pairs")
print(f"Revenue at break-even: ${130 * breakeven_x:,.2f}")
```

2)	Jill invests a total of $17,500 in three products. She invests one part in a mutual fund which has an annual return of 11%. She invests the second part in government bonds at 7% per year. The third part she puts in CDs at 5% per year. She invests twice as much in the mutual fund as in the CDs. In the first year Jill’s investments bring a total return of $1495. How much did she invest in each product? 

Let

x = mutual fund

y = government bonds

z = CDs

x + y + z = 17500

x = 2z

.11x + .07y + .05z = 1495

```python
from sympy import symbols, solve

x, y, z = symbols('x y z')
solution = solve([x + y + z - 17500,
                  x - 2*z,
                  0.11*x + 0.07*y + 0.05*z - 1495], [x, y, z])
print(f"Mutual fund: ${solution[x]:,.2f}")
print(f"Government bonds: ${solution[y]:,.2f}")
print(f"CDs: ${solution[z]:,.2f}")
```
