# Probability, Bayes' Theorem, Counting and Expected Value

```python
# Probabilities from a contingency table of 74 students (36 male, 24 earned an A)
males = 36
total = 74

p_male = males/total
print(f"Probability of a male is {p_male:.4}")

a = 24
p_a = a/total
print(f"Probability of an A is {p_a:.4}")
```

```python
# Probability of independent events: 6 multiple-choice questions with 5 choices each, answered at random
p_succ = .20
q_count = 6

p_all_correct = p_succ ** q_count
p_at_least_one_wrong = 1 - p_all_correct

print(f"Answering all 6 questions at random gives a probability of {p_at_least_one_wrong:.6f} of getting at least one question wrong.")
```

```python
# Bayes' theorem: 62% wear seatbelts; 11% of belted and 50% of unbelted occupants are ejected in a crash

# Prob seatbelts
p_s = .62

# Seatbelts and ejection
p_s_u_e = .11

# No Seatbelts and ejection
p_n_u_e = .50

# Probability of ejection
p_e = (p_s * p_s_u_e) + (1 - p_s) * p_n_u_e

# Probability of non-ejection
p_z = 1 - p_e

# P(A|B) = P(B|A) * P(A) / P(B)
p_se = p_s_u_e * p_s / p_e
print(f"The probability that someone was wearing a seatbelt given that they were ejected is {p_se}")

# Not wearing seatbelt, not ejected
p_z_u_n = 1 - p_n_u_e

# No seatbelt
p_n = 1 - p_s

# P(N | Z) = P(Z | N) * P(N) / P(Z)
p_nz = p_z_u_n * p_n / p_z

print(f"The probability that someone was not wearing a seatbelt given that they were not ejected is {p_nz}")
```

```python
# Counting: plates with 2 numbers, then 3 letters, then 2 numbers (repetition allowed)
n_count = 10
l_count = 26

n = 2
l = 3
n2 = 2

tot_plates = n_count ** n * l_count ** l * n_count ** n2
print(f"There are a total of {tot_plates:,} possible plates with 4 numbers and 3 letters")
```

```python
# Expected value: 0.4% of products fail within the 2 year extended warranty, replacement costs $150, warranty sells for $20
p_failure = .004
cost = 150

warranty_payout = p_failure * cost
warranty_income = 20

value = warranty_income - warranty_payout
r_value = round(value, 2)
print(f"The expected value of each warranty is ${r_value:2}")
```
