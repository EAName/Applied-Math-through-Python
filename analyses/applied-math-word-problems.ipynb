## Three-variable linear system
Solve x + 5y + 7z = -17, x - 2y - 5z = 15, -5x - 4y + 2z = -16.

```python
import numpy as np
from numpy.linalg import inv
```

```python
rhs = [-17, 15, -16]
rhs = np.matrix(rhs)
rhs = np.transpose(rhs)
print('\nRight Hand Side of Equation')
print(rhs)

A = [[1, 5, 7], [1, -2, -5], [-5, -4, 2]]
A = np.matrix(A)
print('\nMatrix A')
print(A)
```

```python
print('\nInverse of A')
IA = inv(A)
print(IA)
```

```python
result = np.dot(IA, rhs)
result = np.round(result, 4)
print('\n Solution to Problem')
print(result)
```

## Product mix (maximize profit)
Miguel manages a store that sells MP3 players. Each MX300 takes 4 hours to create the plastic pieces, 2 hours to create the electronics, and 18 hours of labor to assemble the finished product. Each Jammer requires 2 hours to create the plastic pieces, 4 hours to create the electronics, and 16 hours of labor to assemble the finished product. The factory has 440 hours to create the plastic pieces, 480 hours to create the electronics, and 2120 hours of labor to assemble the finished product each week. If each MX300 generates \$7 in profit, and each Jammer generates \$15, how many of each of the MP3 players should Miguel have the store sell each week to maximize profit?

```python
from pulp import LpVariable, LpProblem, LpMaximize, LpStatus, value, LpMinimize 

# declare your variables
x1 = LpVariable("x1", 0, None) # x1>=0
x2 = LpVariable("x2", 0, None) # x2>=0




# defines the problem
prob = LpProblem("problem", LpMaximize)

# defines the constraints
prob += 4*x1 + 2*x2 <= 440
prob += 2*x1 + 4*x2 <= 480
prob += 18*x1 + 16*x2 <= 2120

# defines the objective function to maximize
prob += 7*x1 + 15*x2

# solve the problem
status = prob.solve(p.PULP_CBC_CMD(msg=0))
LpStatus[status]

# print the results
print("Pulp Solution for x1, x2")
print(value(x1))
print(value(x2))
# print the results (integer values)
print(int(value(x1)),int(value(x2)))
print("The maximum profit is $",(7*value(x1)) + 15*(value(x2)))
```

## Orchard scheduling (minimize cost)
Three orchards produce Fuji, Pink Lady and Honeycrisp apples. Per hour of operation, Orchard 1 yields 25 / 0 / 10 bushels, Orchard 2 yields 40 / 25 / 0 and Orchard 3 yields 20 / 20 / 25. At least 1910 bushels of Fuji, 950 of Pink Lady and 450 of Honeycrisp are required. Hourly costs are \$710, \$780 and \$710. How many hours should each orchard operate to minimize cost?

```python
from pulp import LpVariable, LpProblem, LpMaximize, LpStatus, value, LpMinimize 

# declare your variables (hours of operation)
x1 = LpVariable("x1", 0, None) # Orchard 1
x2 = LpVariable("x2", 0, None) # Orchard 2
x3 = LpVariable("x3", 0, None) # Orchard 3

# defines the problem
prob = LpProblem("problem", LpMinimize)

# defines the constraints (bushels required)
prob += 25*x1 + 40*x2 + 20*x3 >= 1910  # Fuji
prob += 25*x2 + 20*x3 >= 950           # Pink Lady
prob += 10*x1 + 25*x3 >= 450           # Honeycrisp

# defines the objective function to minimize
prob += 710*x1 + 780*x2 + 710*x3

# solve the problem
status = prob.solve(p.PULP_CBC_CMD(msg=0))
LpStatus[status]

# print the results
print("Pulp Solution for x1, x2, x3")
print(value(x1))
print(value(x2))
print(value(x3))
print("The minimum cost is $", value(prob.objective))
```

## Bayes' theorem: false negatives
In the general population, only 0.4% have undiagnosed colorectal cancer. The hemoccult test is used in the detection of colorectal cancer is less expensive and much less invasive alternative to colonoscopy. However it is also much less accurate. The test is positive 50% of the time if the patient has the disease, and will be positive 4% of the time if the patient does not have the disease. What is the probability that a patient has colorectal cancer given that the test is negative? Round your answer to 4 decimal places as needed.


We are given here that: P( cancer ) = 0.004

Also, we are given that: P( positive | cancer ) = 0.5, therefore P( negative | cancer ) = 1 - 0.50 = 0.50

P( positive | no cancer ) = 0.04, therefore P( negative | no cancer) = 1 - 0.04 = 0.96

Using law of total probability, we get here:

P( negative ) = P( negative | cancer )P( cancer ) + P( negative | no cancer)P( no cancer )

P( negative ) = 0.50*0.004 + 0.96*(1 - 0.004) = 0.95816

Now using bayes theorem, the probability of false negative is computed here as:

P( cancer | negative ) = P( negative and cancer ) / P( negative ) = 0.5*0.004/0.95816 = 0.0021

Therefore 0.0021 is the required probability here.

```python
def p_disease_given_negative(prevalence, p_pos_given_disease, p_pos_given_healthy):
    """P(disease | negative test) by Bayes' theorem."""
    p_neg_given_disease = 1 - p_pos_given_disease
    p_neg_given_healthy = 1 - p_pos_given_healthy
    p_neg = p_neg_given_disease*prevalence + p_neg_given_healthy*(1 - prevalence)
    return p_neg_given_disease*prevalence / p_neg
```

```python
p = p_disease_given_negative(prevalence=0.004, p_pos_given_disease=0.50, p_pos_given_healthy=0.04)
print("Probability of the test-taker having colorectal cancer given the test being negative is:",round(p,4))
```

## Marginal cost
The cost (in hundreds of dollars) to produce x units of a product is (7x - 8)/(3x² + x). Find the marginal cost to produce 7 units.

```python
from sympy import Symbol, Derivative, exp, lambdify

x = Symbol('x')
cost = (7*x - 8)/(3*x**2 + x)
marginal = Derivative(cost, x).doit()
print(f"Marginal cost at x = 7: ${float(marginal.subs(x, 7))*100:.2f}")
```

## Absolute minimum of an exchange rate
The value of the U.S. dollar (in Canadian dollars) between 2010 and 2014 can be approximated by

f(t) = -0.002693t³ + 0.03614t² - 0.1005t + 1.0647

where t is the number of years since 2010. In what year did the value reach its absolute minimum, and what was the minimum value?

```python
from sympy import Symbol, solve, Derivative
x = Symbol('x')
f = -0.002693*x**3 + 0.03614*x**2 - 0.1005*x + 1.0647
d1 = Derivative(f, x).doit()
d1
```

```python
critical_points = solve(d1)
critical_points
```

```python
# Compare the critical points inside [0, 4] with the endpoints
candidates = [0, 4] + [c for c in critical_points if 0 <= c <= 4]
values = {round(float(c), 4): round(float(f.subs(x, c)), 4) for c in candidates}
print(values)
t_min = min(values, key=values.get)
print(f"Absolute minimum at t = {t_min}, during {2010 + int(t_min)}, value = ${values[t_min]:.2f}")
```

## Oil leak
An oil tanker is leaking oil at a rate given in barrels per hour by L'(t) = 72 ln(t+1)/(t+1), where t is hours after the tanker hits a rock.

Integrate: L(t) = ∫ 72 ln(t+1)/(t+1) dt = 72(ln(t+1))²/2 + c

When t = 0, L(0) = 0, so L(t) = 36(ln(t+1))²

First day, t = 24: L(24) = 36(ln(25))² ≈ 373

Second day: L(48) - L(24) = 36(ln(49))² - 36(ln(25))² ≈ 172

```python
import math

L = lambda t: 36 * math.log(t + 1)**2
print(f"First day: {L(24):.0f} barrels")
print(f"Second day: {L(48) - L(24):.0f} barrels")
```

## Total profit from a rate of profit
A small company's rate of profit (in thousands of dollars) after t years is P'(t) = (4t + 4)(t² + 2t + 2)^(1/3). Find the total profit in the first three years and the profit in the fourth year.

With u = t² + 2t + 2, du = (2t + 2)dt, so ∫P'(t)dt = 2∫u^(1/3)du = 1.5u^(4/3).

```python
from sympy import Symbol, integrate, Rational

t = Symbol('t')
P_prime = (4*t + 4)*(t**2 + 2*t + 2)**Rational(1, 3)
first_three = integrate(P_prime, (t, 0, 3))
fourth_year = integrate(P_prime, (t, 3, 4))
print(f"Profit in the first three years: ${float(first_three)*1000:,.0f}")
print(f"Profit in the fourth year: ${float(fourth_year)*1000:,.0f}")
```

## Uniform distribution
The life span of a certain insect (in days) is uniformly distributed over the interval [22, 38].

For a uniform distribution, f(x) = 1/(b - a) for x in [a, b], so with a = 22 and b = 38, f(x) = 1/16.

The expected life of the insect is μ = (a + b)/2 = 0.5(22 + 38) = 30 days.

The probability that one of these insects, randomly selected, lives longer than 32 days is (38 - 32)/16 = 0.375.

## Constrained cost minimization
The total cost to produce x large jewelry-making kits and y small ones is C(x, y) = 2x² + 6y² + 4xy + 10. If a total of ten kits must be made, how should production be allocated so that total cost is minimized? What is the minimum cost?

```python
from sympy import symbols, Derivative
from sympy.solvers import solve 

x, y = symbols('x y')
# x + y = 10
x = 10 - y

# Substitute x = 10 - y into the cost function
C = 2*x**2 + 6*y**2 + 4*x*y + 10
dC = Derivative(C, y).doit()

# The cost is minimized where dC = 0, or at an endpoint of 0 <= y <= 10
candidates = [0, 10] + [v for v in solve(dC, y) if 0 <= v <= 10]
y_value = min(candidates, key=lambda v: C.subs({y: v}))
print("y =", y_value)

x_value = x.subs({y:y_value})
print("x =", x_value)

C_value = C.subs({y:y_value})
print("Minimal costs will be $",C_value)
```
