<div class="alert alert-block alert-success"><b>Problem 2:</b> Do the lines x + 8y = 70 and 4x + 5y = 74 intersect at the point (6, 8)?</div>



```python
import sympy as sym
# Use sympy package to display the equation
sym.init_printing()
#sym.solve()
x,y = sym.symbols('x,y')

solns = sym.solve([
    1*x + 8*y - 70, 
    4*x + 5*y - 74],
    [x, y])
print(f"The lines intersect at ({solns[x]}, {solns[y]}), so they do not intersect at (6, 8).")
```

<div class="alert alert-block alert-success"><b>Problem 3: </b>Solve the system x + y = -4, -3x + 3y = -6 by graphing.</div>

```python
import sympy as sym
import numpy as np
import matplotlib.pyplot as plt
# Use sympy package to display the equation
sym.init_printing()
x,y = sym.symbols('x,y')

solns = sym.solve([
    1*x + 1*y + 4,   # Set equations equal to 0 for solving
    -3*x + 3*y + 6],
    [x, y])
print(f"Intersection is ({solns[x]}, {solns[y]}) ")


# Eq 1
# x + y = -4
# y = -x + -4 
# Eq 2
# -3x + 3y = - 6
# 3y = 3x - 6
# y = x - 2

x_vals = np.arange(-5, 6)
y1 = -1 * x_vals + -4  # Eq 1
y2 = x_vals - 2  # Eq 2

plt.plot(x_vals, y1)
plt.plot(x_vals, y2)
plt.scatter(solns[x], solns[y], color='k')
plt.legend(['x + y = -4', '-3x + 3y = -6'])
plt.show()
```

<div class="alert alert-block alert-success"><b>Problem 5: </b>A jar holds 1,254 pennies weighing 3,505.2 g in total. Old pennies weigh 3.1 g and new pennies weigh 2.5 g. How many of each are there?
</div>

```python
import sympy as sym
# Use sympy package to display the equation
sym.init_printing()

# X is old pennies, Y is new pennies
x,y = sym.symbols('x,y')

solns = sym.solve([
    3.1*x + 2.5*y - 3505.2,   # weight
    x + y - 1254],  # count
    [x, y])
print(f"There are {int(solns[x])} old pennies and {int(solns[y])} new pennies")
```

<div class="alert alert-block alert-success"><b>Problem 6:</b> Ken ate 5 tacos and burritos with 1,850 calories in total. A taco has 190 calories and a burrito has 490. How many of each did he eat?</div>

```python
import sympy as sym
# Use sympy package to display the equation
sym.init_printing()

# X is tacos, Y is burritos
x,y = sym.symbols('x,y')

solns = sym.solve([
    190 * x + 490 * y - 1850,   # calories
    x + y - 5],  # count
    [x, y])
print(f"Ken ate {solns[x]} tacos and {solns[y]} burritos")
```

<div class="alert alert-block alert-success"><b>Problem 7:</b> Roberto sold 10 small and 20 large boxes for \$455; Rosa sold 10 small and 16 large boxes for \$381. What does each box size cost?</div>

```python
from numpy import matrix, transpose
from numpy import linalg
M= [[10, 20], # Roberto's sales
    [10, 16]]  # Rosa's sales
M= matrix(M)
O= [455,  # Roberto total
    381]  # Rosa total
O= transpose(matrix(O)) 
result= linalg.solve(M,O)
print ('\nSmall box and large box costs\n') 
print (result)
```

<div class="alert alert-block alert-success"><b>Problem 8:</b> Solve x - 2y + z = -2, y + 2z = 5, x + y + 3z = 5.</div>

```python
import sympy as sym
# Use sympy package to display the equation
sym.init_printing()
x,y,z = sym.symbols('x,y,z')

solns = sym.solve([
    x - 2 * y + z + 2,
    y + 2 * z - 5,
    x + y + 3 * z - 5],
    [x, y, z])
print(f"x: {solns[x]}, y: {solns[y]}, z: {solns[z]}")
```

<div class="alert alert-block alert-success"><b>Problem 9:</b> Solve the 3 x 3 system with the inverse matrix.</div>

```python
rhs = [28000, 750000, 400000]
rhs = np.matrix(rhs)
rhs = np.transpose(rhs)
print('\nRight Hand Side of Equation')
print(rhs)

A = [[1, 1, 1],
     [48, 36, 24],
     [24, 36, 12]]
A = np.matrix(A)
print('\nMatrix A')
print(A)
IA = np.linalg.inv(A)

result = np.dot(IA, rhs)
result = np.rint(result)
result
```

<div class="alert alert-block alert-success"><b>Problem 10:</b> Blend composition matrix times demand matrix, then the cost of the raw materials.</div>

```python
blends = [[.65, .6, .95], 
          [.35, .4, .05]]
A = np.matrix(blends)
print('\nMatrix A')
print(A)


print('\nMatrix B')
demands = [[9, 10],
           [2, 3],
           [33, 33]]

B = np.matrix(demands)
print(B)

prod = np.dot(A, B)
print('\nProduct')
print(prod)

# Because these are in thousands of pounds
thous_prod = 1000 * prod

prices = [3.33, .73]
cost = np.dot(prices, thous_prod)
print("\nCost")
print(cost) 
```

```python
markups = [[.07, 0, 0, 0],[0, .13, 0, 0], [0, 0, .11, 0], [0, 0 , 0, .22]]
A = np.matrix(markups)
print('\nMatrix A')
print(A)


print('\nMatrix B')
prices  = [[12, 14],
           [22, 19],
           [27, 29],
           [64, 46]]

B = np.matrix(prices)
print(B)

profit = A * B 
profit_thou = profit * 1000
print('\nProfit')
print(profit_thou)

lux_a_three_prof = profit_thou[3, 0] * 5
print('\nThree Brand A Luxury Car Profits')
print(lux_a_three_prof)
```

<div class="alert alert-block alert-success"><b>Problem 12:</b> Solve 7x - 11y = -2, -x + 3y = 3 with the inverse matrix.</div>

```python
lhs = np.matrix([[7, -11], [-1, 3]])
rhs = [-2, 3]

print("Coefficient Matrix")
print(lhs)

print("Inverse of it is")
inv_lhs = np.linalg.inv(lhs)
print(inv_lhs)

sol = np.dot(inv_lhs, rhs)
print(f"The solutions are {sol}")
```

<div class="alert alert-block alert-success"><b>Problem 14:</b> Solve 3x - 5y = -13, -x - 3y = -19 with Cramer's rule, then a 3 x 3 system with Cramer's rule.</div>

```python
a1 = 3
a2 = -1
b1 = -5
b2 = -3
c1 = -13
c2 = -19

D = (a1 * b2) - (a2 * b1)
print(f'D is {D}')

Dx = (c1 * b2) - (c2 * b1)
Dy = (a1 * c2) - (a2 * c1)
print(f'Dx is {Dx}')
print(f'Dy is {Dy}')

x = Dx/D
y = Dy/D
print(f"The solution is ({x}, {y})")
```

```python
# importing Numpy package
import numpy as np
  
# creating a 3X3 Numpy matrix
n_array = np.array([[6, -1, -5],
                    [-4, 1, -2],
                    [3, -3, 4]])

rhs = np.array([-4, 17, -9]) 

x_matrix = np.array([[-4, -1, -5],
               [17, 1, -2],
               [-9, -3, 4]])

y_matrix = np.array([[6, -4, -5],
               [-4, 17, -2],
               [3, -9, 4]])

z_matrix = np.array([[6, -1, -4],
               [-4, 1, 17],
               [3, -3, -9]])
# Displaying the Matrix
print("Numpy Matrix is:")
print(n_array)
  
# calculating the determinant of matrix
det = np.linalg.det(n_array)
Dx = np.linalg.det(x_matrix) 
Dy = np.linalg.det(y_matrix) 
Dz = np.linalg.det(z_matrix) 
print("\nDeterminant of given 3X3 square matrix:")
print(round(det))
print(Dx)
print(Dy)
print(Dz) 
print(x_matrix)
print(y_matrix)
print(z_matrix)
print(Dx/det)
print(Dy/det)
print(Dz/det)
```
