# Derivatives, Concavity and Optimization

```python
# Relative extrema of P(x) = 2x^3 + 18x^2 + 30x - 5

from sympy import Symbol, solveset, lambdify, Interval, oo  # oo is sympys representation of Infinity
from sympy.plotting import plot
from math import floor

x = Symbol('x')  # Set a symbol
P = 2*x**3 + 18*x**2 + 30*x -5   # Function to analyze
P_fun = lambdify(x, P)
der_P = P.diff(x)  # derivitive of F

dRoots = solveset(der_P, x, Interval(-9, 9)) # Find the roots of the derivative where x is between -9 and 9
y_vals = [P_fun(r) for r in dRoots]
print(f"Extreme values are {dRoots} at y values of {y_vals} respectively") 
```

```python
# Derivative of f(x) = 8 sqrt(x) + 4
import sympy

x = sympy.Symbol('x')
f = 8 * sympy.sqrt(x) + 4
f_prime = f.diff(x)

f_prime_plot = sympy.plotting.plot(f_prime)
```

```python
# Second derivative test for g(x) = 4x^3 - 36x^2 + 96x at x = 2
from sympy import Symbol, lambdify, plotting

x = Symbol('x')
g = 4 * x **3 - 36 * x**2 + 96 * x
g_first = g.diff(x)
g_second = g_first.diff(x)

print(f"First derivative is {g_first}")
print(f"Second derivative is {g_second}")

g_first_f = lambdify(x, g_first)
g_second_f = lambdify(x, g_second)

p1 = g_second_f(2)

print(f"g''(2) = {p1}")

gp = plotting.plot(g, show=False, ylim=(-1000, 1000), line_color='b', label='g', legend=True)
gp1 = plotting.plot(g_first, show=False, line_color='r', label="g'")
gp2 = plotting.plot(g_second, show=False, line_color='g', label="g''")

gp.append(gp1[0])
gp.append(gp2[0])
gp.show()

print("We can visually verify our interpretation of concave down and a maxima")
```

```python
# Maximum of S(t) = 170t / (t + 4)^2, sales in thousands of dollars
from sympy import Symbol, lambdify, solve, plotting, oo, solveset, Interval

t = Symbol('t')
S = (170 * t) / (t + 4)**2

S_prime = S.diff(t)

print("Sales Plot")
plotting.plot(S, xlim=(0, 15), ylim=(0, 100))


dRoots = solveset(S_prime, t, Interval(0, oo)) # Find the roots of the derivative where t is between 0 and infinity
print(f"Extreme values are {dRoots}") # gives a set with a single extrema, this is our answer

S_func = lambdify(t, S)  # Turn the symbolic equation into a callable function so we can find the S(t) for our found t value
max_val = S_func(dRoots.args[0])
print(f"Maximum sales are {max_val * 1000} and occurs at {dRoots.args[0]}")
```
